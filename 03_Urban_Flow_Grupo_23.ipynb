# Ejercicio 04

> Puntos: 1

Completar la función `procesar_fila_csv` que recibe como parámetro un `dict` y debe cargar los datos de una fila del archivo `data/processed/speeding_fines_image.csv` y mapearla según las clases definidas en el punto anterior.

Se implementan las clases correspondientes al modelo lógico definido en el ejercicio anterior.

Luego se desarrolla la función `procesar_fila_csv`, que recibe una fila del archivo `speeding_fines_image.csv` en formato diccionario y la transforma en objetos de dominio.

````python
#@title Clases del modelo lógico

class Vehiculo:
    def __init__(self, patente):
        self.patente = patente


class Radar:
    def __init__(self, radar_id, ubicacion):
        self.radar_id = radar_id
        self.ubicacion = ubicacion


class Evidencia:
    def __init__(self, imagen, patente_imagen, ratio):
        self.imagen = imagen
        self.archivo_imagen = imagen
        self.patente_imagen = patente_imagen
        self.ratio = ratio


class Multa:
    def __init__(
        self,
        multa_id,
        fecha,
        hora,
        velocidad_registrada,
        velocidad_maxima,
        estado_multa,
        vehiculo,
        radar,
        evidencia=None,
        exceso_velocidad_real=None,
        exceso_velocidad=None
    ):
        self.multa_id = multa_id
        self.fecha = fecha
        self.hora = hora
        self.velocidad_registrada = velocidad_registrada
        self.velocidad_maxima = velocidad_maxima
        self.estado_multa = estado_multa
        self.vehiculo = vehiculo
        self.radar = radar
        self.evidencia = evidencia
        self.exceso_velocidad_real = exceso_velocidad_real
        self.exceso_velocidad = exceso_velocidad
````

````python
#@title Función procesar_fila_csv

def procesar_fila_csv(row: dict) -> Multa:
    vehiculo = Vehiculo(
        patente=row["patente"]
    )

    radar = Radar(
        radar_id=row["radar_id"],
        ubicacion=row["ubicacion"]
    )

    evidencia = None

    if pd.notna(row.get("imagen")) and row.get("imagen") != "":
        evidencia = Evidencia(
            imagen=row["imagen"],
            patente_imagen=row.get("patente_imagen"),
            ratio=row.get("ratio")
        )

    multa = Multa(
        multa_id=row["multa_id"],
        fecha=row["fecha"],
        hora=row["hora"],
        velocidad_registrada=row["velocidad_registrada"],
        velocidad_maxima=row["velocidad_maxima"],
        estado_multa=row["estado_multa"],
        vehiculo=vehiculo,
        radar=radar,
        evidencia=evidencia,
        exceso_velocidad_real=row.get("exceso_velocidad_real"),
        exceso_velocidad=row.get("exceso_velocidad")
    )

    return multa
````

````python
#@title Validación de la función

row_ejemplo = {
  'multa_id': 'M-999',
  'patente': 'AA123BB',
  'fecha': '2023-10-25',
  'hora': '14:30:00',
  'velocidad_registrada': 130,
  'velocidad_maxima': 110,
  'ubicacion': 'Au. 25 de Mayo Km 4',
  'radar_id': 'R-01',
  'estado_multa': 'IMPAGA',
  'imagen': 'foto_123.jpg',
  'patente_imagen': 'AA12388',
  'ratio': 0.85
}

multa_objeto = procesar_fila_csv(row_ejemplo)

print(f"Multa creada para el auto: {multa_objeto.vehiculo.patente}")
print(f"Generada por el radar en: {multa_objeto.radar.ubicacion}")
if multa_objeto.evidencia:
  print(f"Evidencia adjunta: {multa_objeto.evidencia.archivo_imagen}")
````

Diseñar el modelo relacional (según el modelo de datos creado en el punto 3) que usaremos con el ORM de SQLAlchemy.

Cada modelo debe incluir:

- Relaciones entre tablas

## Modelo relacional

A partir del modelo lógico definido en el ejercicio anterior, se diseñó el modelo relacional utilizando el ORM SQLAlchemy.

### Tablas definidas

#### Vehiculos
Almacena la información de los vehículos infraccionados.

Clave primaria:
- patente

#### Radares
Almacena la información de los radares que registran las infracciones.

Clave primaria:
- radar_id

#### Multas
Almacena los datos de cada infracción registrada.

Clave primaria:
- multa_id

Claves foráneas:
- patente → Vehiculos
- radar_id → Radares

#### Evidencias
Almacena la evidencia visual asociada a una multa.

Clave primaria:
- evidencia_id

Clave foránea:
- multa_id → Multas

### Relaciones

- Un vehículo puede tener muchas multas.
- Un radar puede generar muchas multas.
- Una multa pertenece a un único vehículo.
- Una multa pertenece a un único radar.
- Una multa puede tener una única evidencia asociada.
- Una evidencia pertenece a una única multa.

Para mejorar la legibilidad durante las pruebas y consultas, cada modelo implementa el método `__repr__`.

````python
#@title Modelo relacional con SQLAlchemy

Base = declarative_base()


class VehiculoModel(Base):
    __tablename__ = "vehiculos"

    patente = Column(String, primary_key=True)

    multas = relationship("MultaModel", back_populates="vehiculo")

    def __repr__(self):
        return f"<Vehiculo(patente='{self.patente}')>"


class RadarModel(Base):
    __tablename__ = "radares"

    radar_id = Column(String, primary_key=True)
    ubicacion = Column(String)

    multas = relationship("MultaModel", back_populates="radar")

    def __repr__(self):
        return f"<Radar(radar_id='{self.radar_id}', ubicacion='{self.ubicacion}')>"


class MultaModel(Base):
    __tablename__ = "multas"

    multa_id = Column(String, primary_key=True)
    fecha = Column(String)
    hora = Column(String)
    velocidad_registrada = Column(Float)
    velocidad_maxima = Column(Float)
    estado_multa = Column(String)
    exceso_velocidad_real = Column(Float)
    exceso_velocidad = Column(Float)

    patente = Column(String, ForeignKey("vehiculos.patente"))
    radar_id = Column(String, ForeignKey("radares.radar_id"))

    vehiculo = relationship("VehiculoModel", back_populates="multas")
    radar = relationship("RadarModel", back_populates="multas")
    evidencia = relationship("EvidenciaModel", back_populates="multa", uselist=False)

    def __repr__(self):
        return f"<Multa(multa_id='{self.multa_id}', patente='{self.patente}', radar_id='{self.radar_id}')>"


class EvidenciaModel(Base):
    __tablename__ = "evidencias"

    evidencia_id = Column(Integer, primary_key=True, autoincrement=True)
    multa_id = Column(String, ForeignKey("multas.multa_id"), unique=True)
    imagen = Column(String)
    patente_imagen = Column(String)
    ratio = Column(Float)

    multa = relationship("MultaModel", back_populates="evidencia")

    def __repr__(self):
        return f"<Evidencia(evidencia_id={self.evidencia_id}, multa_id='{self.multa_id}', imagen='{self.imagen}')>"
````

````python
#@title Crear y poblar base de datos transito

csv_multas_path = "urban_flow/data/interim/speeding_fines.csv"
csv_evidencias_path = "urban_flow/data/processed/speeding_fines_image.csv"

engine = create_engine("sqlite:///transito.db")

Base.metadata.drop_all(engine)
Base.metadata.create_all(engine)

Session = sessionmaker(bind=engine)
session = Session()

df_multas = pd.read_csv(csv_multas_path)
df_evidencias = pd.read_csv(csv_evidencias_path)

df_multas["patente"] = df_multas["patente"].fillna("PATENTE_SIN_DATO")
df_multas["radar_id"] = df_multas["radar_id"].fillna("RADAR_SIN_ID")

df_evidencias["patente"] = df_evidencias["patente"].fillna("PATENTE_SIN_DATO")
df_evidencias["radar_id"] = df_evidencias["radar_id"].fillna("RADAR_SIN_ID")

df_multas["multa_id"] = [f"M-{i+1}" for i in range(len(df_multas))]

evidencias_por_clave = {}

for _, row in df_evidencias.iterrows():
    if pd.notna(row.get("imagen")):
        clave = (
            str(row["patente"]),
            str(row["fecha"]),
            str(row["hora"])
        )

        if clave not in evidencias_por_clave:
            evidencias_por_clave[clave] = row

for _, row in df_multas.iterrows():
    vehiculo = session.get(VehiculoModel, row["patente"])

    if vehiculo is None:
        vehiculo = VehiculoModel(
            patente=row["patente"]
        )
        session.add(vehiculo)

    radar = session.get(RadarModel, row["radar_id"])

    if radar is None:
        radar = RadarModel(
            radar_id=row["radar_id"],
            ubicacion=row["ubicacion"]
        )
        session.add(radar)

    multa = MultaModel(
        multa_id=row["multa_id"],
        fecha=row["fecha"],
        hora=row["hora"],
        velocidad_registrada=row["velocidad_registrada"],
        velocidad_maxima=row["velocidad_maxima"],
        estado_multa=row["estado_multa"],
        exceso_velocidad_real=row.get("exceso_velocidad_real"),
        exceso_velocidad=row.get("exceso_velocidad"),
        patente=row["patente"],
        radar_id=row["radar_id"]
    )

    session.add(multa)

    clave_multa = (
        str(row["patente"]),
        str(row["fecha"]),
        str(row["hora"])
    )

    evidencia_row = evidencias_por_clave.get(clave_multa)

    if evidencia_row is not None:
        evidencia = EvidenciaModel(
            multa_id=row["multa_id"],
            imagen=evidencia_row["imagen"],
            patente_imagen=evidencia_row["patente_imagen"],
            ratio=evidencia_row["ratio"]
        )
        session.add(evidencia)

session.commit()
````

## Consultas sobre la base de datos

Se implementan consultas sobre la base de datos `transito` utilizando SQLAlchemy.

El objetivo es obtener información relevante sobre las multas registradas, los vehículos reincidentes, los radares con mayor actividad, las multas sin evidencia asociada y el porcentaje de infracciones confirmadas visualmente.

````python
#@title Top 10 patentes con mayor cantidad de multas

from sqlalchemy import func

top_patentes = (
    session.query(
        MultaModel.patente,
        func.count(MultaModel.multa_id).label("cantidad")
    )
    .group_by(MultaModel.patente)
    .order_by(func.count(MultaModel.multa_id).desc())
    .limit(10)
    .all()
)

print("--- Top 10 patentes con mayor cantidad de multas ---")
print(f"{'Patente':<10} | {'Cantidad':<8}")
print(f"{'-'*10} | {'-'*8}")

for patente, cantidad in top_patentes:
    print(f"{patente:<10} | {cantidad:<8}")
````

````python
#@title Top 10 multas sin evidencia asociada

total_sin_evidencia = (
    session.query(MultaModel)
    .outerjoin(EvidenciaModel)
    .filter(EvidenciaModel.evidencia_id == None)
    .count()
)

multas_sin_evidencia = (
    session.query(MultaModel)
    .outerjoin(EvidenciaModel)
    .filter(EvidenciaModel.evidencia_id == None)
    .order_by(MultaModel.fecha.asc())
    .limit(10)
    .all()
)

print(f"--- Total multas sin evidencia ({total_sin_evidencia}) ---")
print("--- Top 10 ---")
print(f"{'Multa':<10} | {'Fecha':<10}")
print(f"{'-'*10} | {'-'*10}")

for multa in multas_sin_evidencia:
    print(f"{multa.multa_id:<10} | {multa.fecha:<10}")
````

````python
#@title Radares más activos

radares_activos = (
    session.query(
        RadarModel.radar_id,
        RadarModel.ubicacion,
        func.count(MultaModel.multa_id).label("cantidad")
    )
    .join(MultaModel, RadarModel.radar_id == MultaModel.radar_id)
    .group_by(RadarModel.radar_id, RadarModel.ubicacion)
    .order_by(func.count(MultaModel.multa_id).desc())
    .all()
)

print("--- Radares más activos ---")
print(f"{'Radar':<10} | {'Ubicación':<25} | {'Cantidad':<8}")
print(f"{'-'*10} | {'-'*25} | {'-'*8}")

for radar, ubicacion, cantidad in radares_activos:
    print(f"{radar:<10} | {ubicacion:<25} | {cantidad:<8}")
````

````python
#@title Top 10 reincidentes por período

fecha_inicio = "2020-01-01"
fecha_fin = "2020-12-31"

reincidentes = (
    session.query(
        MultaModel.patente,
        func.count(MultaModel.multa_id).label("cantidad")
    )
    .filter(MultaModel.fecha >= fecha_inicio)
    .filter(MultaModel.fecha <= fecha_fin)
    .group_by(MultaModel.patente)
    .order_by(func.count(MultaModel.multa_id).desc())
    .limit(10)
    .all()
)

print(f"--- Top 10 Reincidentes ({fecha_inicio} a {fecha_fin}) ---")
print(f"{'Patente':<10} | {'Cantidad':<8}")
print(f"{'-'*10} | {'-'*8}")

for patente, cantidad in reincidentes:
    print(f"{patente:<10} | {cantidad:<8}")
````

## Base de datos vectorial

En este ejercicio se crea una base de datos vectorial llamada `patente_vectorial` utilizando ChromaDB.

Para generar los vectores se utiliza el modelo base de OpenCLIP. Cada imagen de patente se transforma en un embedding y se almacena junto con el identificador del vehículo correspondiente.

De esta manera, luego será posible consultar la base vectorial con una imagen y recuperar el vehículo más similar.

````python
#@title Recuperar imágenes necesarias para OpenCLIP

if not os.path.exists("urban_flow/data/raw/imgs"):
    print("No se encontraron las imágenes localmente. Se intentan recuperar desde DVC en Sprint_3.")
    !dvc pull urban_flow/data/raw/imgs.dvc
else:
    print("Las imágenes originales ya están disponibles en Sprint_3.")

if not os.path.exists("urban_flow/data/raw/urban_flow_plates.zip"):
    print("No se encontró el ZIP localmente. Se intenta recuperar desde DVC en Sprint_3.")
    !dvc pull urban_flow/data/raw/urban_flow_plates.zip.dvc
else:
    print("El ZIP ya está disponible en Sprint_3.")

if not os.path.exists("urban_flow/data/raw/imgs"):
    raise FileNotFoundError("No se pudieron recuperar las imágenes necesarias para OpenCLIP.")

!find urban_flow/data/raw/imgs -type f | head
````

````python
#@title Crear y poblar base vectorial patente_vectorial

device = "cuda" if torch.cuda.is_available() else "cpu"

modelo_clip, _, preprocess = open_clip.create_model_and_transforms(
    "ViT-B-32",
    pretrained="openai"
)

modelo_clip = modelo_clip.to(device)
modelo_clip.eval()

cliente_chroma = chromadb.PersistentClient(path="./patente_vectorial")

try:
    cliente_chroma.delete_collection("patente_vectorial")
except:
    pass

coleccion_patentes = cliente_chroma.create_collection(
    name="patente_vectorial"
)

df_evidencias_vector = pd.read_csv("urban_flow/data/processed/speeding_fines_image.csv")

df_evidencias_vector = df_evidencias_vector.dropna(subset=["imagen", "patente"])

ids = []
embeddings = []
metadatas = []
documents = []

for index, row in df_evidencias_vector.iterrows():
    imagen_path = row["imagen"]

    if not os.path.exists(imagen_path):
        imagen_path = os.path.join("urban_flow/data/raw/imgs/urban_flow_plates", os.path.basename(str(row["imagen"])))

    if not os.path.exists(imagen_path):
        continue

    try:
        imagen = Image.open(imagen_path).convert("RGB")
        imagen_preprocesada = preprocess(imagen).unsqueeze(0).to(device)

        with torch.no_grad():
            vector = modelo_clip.encode_image(imagen_preprocesada)
            vector = vector / vector.norm(dim=-1, keepdim=True)

        vector = vector.cpu().numpy()[0].tolist()

        vehiculo = session.get(VehiculoModel, row["patente"])

        if vehiculo is not None:
            ids.append(f"img_{index}")
            embeddings.append(vector)
            metadatas.append({
                "vehiculo_id": row["patente"],
                "patente": row["patente"],
                "imagen": imagen_path
            })
            documents.append(str(row["patente"]))

    except Exception as e:
        print(f"No se pudo procesar {imagen_path}: {e}")

if len(ids) > 0:
    coleccion_patentes.add(
        ids=ids,
        embeddings=embeddings,
        metadatas=metadatas,
        documents=documents
    )

print("Base vectorial creada: patente_vectorial")
print("Vectores insertados:", coleccion_patentes.count())
````

## Búsqueda de vehículo por imagen

Se implementa la función `buscar_patente_imagen`, que recibe la ruta de una imagen, genera su embedding con OpenCLIP y consulta la base vectorial `patente_vectorial`.

A partir del resultado más similar, se obtiene el identificador del vehículo y se consulta la base relacional para retornar sus datos y multas asociadas.

````python
#@title Función buscar_patente_imagen

def buscar_patente_imagen(imagen_path: str):
    if not os.path.exists(imagen_path):
        return {
            "error": f"No existe la imagen: {imagen_path}"
        }

    imagen = Image.open(imagen_path).convert("RGB")
    imagen_preprocesada = preprocess(imagen).unsqueeze(0).to(device)

    with torch.no_grad():
        vector = modelo_clip.encode_image(imagen_preprocesada)
        vector = vector / vector.norm(dim=-1, keepdim=True)

    vector = vector.cpu().numpy()[0].tolist()

    resultado = coleccion_patentes.query(
        query_embeddings=[vector],
        n_results=1
    )

    if len(resultado["metadatas"][0]) == 0:
        return {
            "error": "No se encontraron coincidencias en la base vectorial."
        }

    metadata = resultado["metadatas"][0][0]
    patente = metadata["vehiculo_id"]

    vehiculo = session.get(VehiculoModel, patente)

    if vehiculo is None:
        return {
            "error": f"No se encontró el vehículo {patente} en la base relacional."
        }

    multas = (
        session.query(MultaModel)
        .filter(MultaModel.patente == patente)
        .all()
    )

    return {
        "patente": vehiculo.patente,
        "cantidad_multas": len(multas),
        "multas": [
            {
                "multa_id": multa.multa_id,
                "fecha": multa.fecha,
                "hora": multa.hora,
                "estado": multa.estado_multa,
                "radar_id": multa.radar_id
            }
            for multa in multas
        ],
        "imagen_mas_similar": metadata["imagen"]
    }
````
