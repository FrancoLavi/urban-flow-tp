# Ejercicio 02 - Verificación del dataset de imágenes

Listar imágenes y tamaños

````python
imgs_directory = "urban_flow/data/raw/imgs"

image_extensions = (
  ".jpg",
  ".jpeg",
  ".png"
)

image_files = []

for root, _, files in os.walk(imgs_directory):
  for file_name in files:
    if file_name.lower().endswith(image_extensions):
      image_path = os.path.join(root, file_name)
      image_size_kb = os.path.getsize(image_path) / 1024

      image_files.append({
        "filename": file_name,
        "size_kb": round(image_size_kb, 2),
        "path": image_path
      })

images_df = pd.DataFrame(image_files)

images_df[["filename", "size_kb"]]
````

##Crear diccionario group_images

````python
group_images = {
  "plates": [],
  "completes": []
}

image_data = []

for image_file in image_files:
  image = cv.imread(image_file["path"])

  if image is not None:
    height, width = image.shape[:2]
    area = width * height
    file_name = image_file["filename"]
    patent = os.path.splitext(file_name)[0].split("_")[0]

    image_data.append({
      "filename": file_name,
      "width": width,
      "height": height,
      "area": area,
      "path": image_file["path"],
      "patent": patent
    })

average_area = sum(
  image["area"] for image in image_data
) / len(image_data)

for image in image_data:
  if image["area"] < average_area:
    group_images["plates"].append(image)
  else:
    group_images["completes"].append(image)

print(
  f"El grupo 'plates' contiene "
  f"{len(group_images['plates'])} imágenes."
)

print(
  f"El grupo 'completes' contiene "
  f"{len(group_images['completes'])} imágenes."
)
````

##Guardar JSON

````python
interim_directory = "urban_flow/data/interim"

os.makedirs(
  interim_directory,
  exist_ok=True
)

group_images_path = (
  f"{interim_directory}/group_images.json"
)

with open(group_images_path, "w", encoding="utf-8") as json_file:
  json.dump(
    group_images,
    json_file,
    indent=2,
    ensure_ascii=False
  )

print(
  "Archivo group_images.json guardado correctamente."
)
````

##Mostrar imágenes aleatorias

````python
def show_random_images(
  group_images: dict
) -> None:

  selected_images = (
    random.sample(
      group_images["plates"],
      4
    )
    + random.sample(
      group_images["completes"],
      4
    )
  )

  figure, axes = plt.subplots(
    4,
    2,
    figsize=(20, 32)
  )

  axes = axes.flatten()

  for index, image_info in enumerate(selected_images):

    image = cv.imread(
      image_info["path"]
    )

    image = cv.cvtColor(
      image,
      cv.COLOR_BGR2RGB
    )

    axes[index].imshow(image)

    axes[index].set_title(
      image_info["filename"]
    )

    axes[index].axis("off")

  plt.tight_layout()
  plt.show()
````

# Ejercicio 03 - Procesamiento de imágenes


##Funciones reutilizables

````python
def create_group_directories(
  base_directory: str
) -> None:

  for group_name in group_images.keys():
    os.makedirs(
      f"{base_directory}/{group_name}",
      exist_ok=True
    )


def show_processed_images(
  group_images: dict[
    str,
    list[dict[str, str | int]]
  ],
  base_directory: str
) -> None:

  selected_images = (
    random.sample(
      group_images["plates"],
      4
    )
    + random.sample(
      group_images["completes"],
      4
    )
  )

  figure, axes = plt.subplots(
    4,
    2,
    figsize=(20, 32)
  )

  axes = axes.flatten()

  for index, image_info in enumerate(selected_images):

    image_path = (
      f"{base_directory}/"
      f"{image_info['group']}/"
      f"{image_info['filename']}"
    )

    image = cv.imread(
      image_path,
      cv.IMREAD_GRAYSCALE
    )

    axes[index].imshow(
      image,
      cmap="gray"
    )

    axes[index].set_title(
      image_info["filename"]
    )

    axes[index].axis("off")

  plt.tight_layout()
  plt.show()
````

##Conversión a escala de grises

````python
gray_scale_directory = (
  "urban_flow/data/interim/imgs/03_01_gray_scale"
)

create_group_directories(
  gray_scale_directory
)

for group_name, images in group_images.items():

  for image_info in images:

    image = cv.imread(
      image_info["path"]
    )

    gray_image = cv.cvtColor(
      image,
      cv.COLOR_BGR2GRAY
    )

    output_path = (
      f"{gray_scale_directory}/"
      f"{group_name}/"
      f"{image_info['filename']}"
    )

    cv.imwrite(
      output_path,
      gray_image
    )

    image_info["group"] = group_name

print(
  "Imágenes convertidas a escala de grises correctamente."
)
````

##Suavizado de imágenes

````python
blur_directory = (
  "urban_flow/data/interim/imgs/03_02_blur"
)

create_group_directories(
  blur_directory
)

for group_name, images in group_images.items():

  for image_info in images:

    image_path = (
      f"{gray_scale_directory}/"
      f"{group_name}/"
      f"{image_info['filename']}"
    )

    gray_image = cv.imread(
      image_path,
      cv.IMREAD_GRAYSCALE
    )

    blur_image = cv.GaussianBlur(
      gray_image,
      (5, 5),
      0
    )

    output_path = (
      f"{blur_directory}/"
      f"{group_name}/"
      f"{image_info['filename']}"
    )

    cv.imwrite(
      output_path,
      blur_image
    )

print(
  "Imágenes suavizadas correctamente."
)
````

Vamos autilizar la función `extraer_patente` para obtener las patentes de las imágenes. Esta patente se almacena en la key `patent` del diccionario `group_images`.

Una vez extraidas las patentes se debe relacionar lo extraido con el dataset previamente tratado y almacenado en `urban_flow/data/interim/speeding_fines.csv`, para finalmente guardar el dataset en `urban_flow/data/processed/speeding_fines_image.csv`.

````python
#@title Definición de la función extraer_patente

import easyocr
def extraer_patente(image_path) -> str:
  """ Extrae la patente de una imagen.

  Args:
      image_path (str): Ruta de la imagen.

  Returns:
      str: Patente extraída.
  """
  reader = easyocr.Reader(['es'])
  detection:list = reader.readtext(image_path)
  if len(detection) == 0:
    return ''
  patent = ''
  for d in detection:
    if isinstance(d,tuple) and len(d) == 3:
      if isinstance(d[1],str):
        patent+=d[1]
  return patent
````

````python
def limpiar_patente(texto):
  if pd.isna(texto):
    return ""

  texto = str(texto).upper()
  texto = re.sub(r"[^A-Z0-9]", "", texto)

  return texto
````

````python
def calcular_ratio(patente_original, patente_detectada):
  original = limpiar_patente(patente_original)
  detectada = limpiar_patente(patente_detectada)

  if len(original) == 0 or len(detectada) == 0:
    return 0

  coincidencias = 0
  largo = min(len(original), len(detectada))

  for i in range(largo):
    if original[i] == detectada[i]:
      coincidencias += 1

  return coincidencias / len(original)
````

````python
df = pd.read_csv("urban_flow/data/interim/speeding_fines.csv")

with open(
    "urban_flow/data/interim/group_images.json",
    "r",
    encoding="utf-8"
) as file:

  group_images = json.load(file)
````

````python
for key in group_images:
  for image_data in group_images[key]:
    image_path = image_data["path"]

    patent = extraer_patente(image_path)

    image_data["patent"] = limpiar_patente(patent)
````

````python
filas_con_imagen = []
imagenes_con_match = set()

for key in group_images:
  for image_data in group_images[key]:

    image_path = image_data["path"]
    patente_imagen = image_data["patent"]

    if len(patente_imagen) < 6:
      continue

    for index, row in df.iterrows():

      patente_original = row["patente"]

      ratio = calcular_ratio(
          patente_original,
          patente_imagen
      )

      if ratio >= 0.80:

        nueva_fila = row.copy()

        nueva_fila["imagen"] = image_path
        nueva_fila["patente_imagen"] = patente_imagen
        nueva_fila["ratio"] = round(ratio * 100, 2)

        filas_con_imagen.append(nueva_fila)

        imagenes_con_match.add(image_path)
````

````python
df_imagenes = pd.DataFrame(filas_con_imagen)

df_imagenes.head()
````

````python
os.makedirs(
    "urban_flow/data/processed",
    exist_ok=True
)

df_imagenes.to_csv(
    "urban_flow/data/processed/speeding_fines_image.csv",
    index=False
)
````
